```python
principal = 10000
rate = 5
time = 2
interest = (principal * rate * time) / 100
total = principal + interest
print("Principal:", principal, "KES")
print("Interest rate:", rate, "% per year")
print("Time:", time, "years")
print("Interest earned:", interest, "KES")
print("Total amount:", total, "KES")
```

```python
# Credit Scoring System
monthly_income = 50000
existing_loans = 10000
employment_status = "Employed"

print("--- Credit Score Assessment ---")
print("Monthly Income: KES", monthly_income)
print("Existing Loan Payments: KES", existing_loans)
print("Employment Status:", employment_status)
print("")

# Check employment first
if employment_status == "Unemployed":
    print("Decision: Loan Rejected - Unemployed")
else:
    ratio = existing_loans / monthly_income
    print(f"Debt-to-income ratio: {ratio*100:.1f}%")
    
    if ratio < 0.30:
        print("Decision: Loan Approved")
    elif ratio <= 0.50:
        print("Decision: Need More Information")
    else:
        print("Decision: Loan Rejected - High debt ratio")
```
